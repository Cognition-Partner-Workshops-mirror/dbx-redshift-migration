```python
import io

import boto3
import pandas as pd

export_path = "s3://retail-finance-delivery-usw2/finance_export/monthly/"
s3 = boto3.client("s3")
bucket, prefix = export_path.removeprefix("s3://").split("/", 1)
pages = s3.get_paginator("list_objects_v2").paginate(Bucket=bucket, Prefix=prefix)
frames = []
for page in pages:
    for item in page.get("Contents", []):
        if item["Key"].endswith(".csv"):
            body = s3.get_object(Bucket=bucket, Key=item["Key"])["Body"].read()
            frames.append(pd.read_csv(io.BytesIO(body)))
monthly_finance = pd.concat(frames, ignore_index=True)
monthly_finance.head()
```
